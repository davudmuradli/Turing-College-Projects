Treating outliers.

```python
numeric_cols = spotify.select_dtypes(include=['float64', 'int64']).columns

Q1 = spotify['duration_ms'].quantile(0.25)
Q3 = spotify['duration_ms'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = spotify[(spotify['duration_ms'] < lower) | (spotify['duration_ms'] > upper)]

for col in numeric_cols:
    Q1 = spotify[col].quantile(0.25)
    Q3 = spotify[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = spotify[(spotify[col] < lower) | (spotify[col] > upper)]

    print(f"Column: {col}, Outliers found: {len(outliers)}")

```

Which features are strongly positively correlated?

```python
corr = spotify.corr(numeric_only=True) #find all columns with numbers
corr_unstacked = corr.unstack() # to turn matrix into a long list of pairs
strong_pos = corr_unstacked[(corr_unstacked > 0.5) & (corr_unstacked < 1)] #if not 1 then it would show pairs like energy energy, loudness loudness
strong_pos.sort_values()
```

Which features are strongly negatively correlated?

```python
weak_pos = corr_unstacked[(corr_unstacked < -0.5)] 
weak_pos.sort_values()
```

Which features are not correlated?

```python
upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
clean_corr = upper.unstack().dropna()

weak_pos = clean_corr[(corr_unstacked < 0.1) & (corr_unstacked > -0.1)]
weak_pos
```

How does the danceability score compare between Pop, Hip-Hop/Rap, Dance/Electronic, and Alternative/Indie genres?

```python
genres = ['Pop', 'Hip-Hop/Rap', 'Dance/Electronic', 'Alternative/Indie']
spotify[spotify['genre'].isin(genres)].groupby('genre').agg(mean_danceability=('danceability', 'mean'))
```

How does the loudness score compare between Pop, Hip-Hop/Rap, Dance/Electronic, and Alternative/Indie genres?

```python
genres = ['Pop', 'Hip-Hop/Rap', 'Dance/Electronic', 'Alternative/Indie']
spotify[spotify['genre'].isin(genres)].groupby('genre').agg(mean_loudness =('loudness', 'mean'))
```
